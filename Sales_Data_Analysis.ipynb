```python
# ============================================
# SALES DATA ANALYSIS — STEPS 6 TO 14
# ============================================

# 6. Customer Segment Analysis
segment_sales = df.groupby('Segment')['Sales'].sum().sort_values(ascending=False)
segment_profit = df.groupby('Segment')['Profit'].sum().sort_values(ascending=False)

print("===== SALES BY SEGMENT =====")
print(segment_sales)

print("\n===== PROFIT BY SEGMENT =====")
print(segment_profit)


# 7. Monthly Sales Trend
df['Month'] = df['Order Date'].dt.to_period('M')

monthly_sales = df.groupby('Month')['Sales'].sum()
monthly_sales.index = monthly_sales.index.to_timestamp()

print("\n===== MONTHLY SALES =====")
print(monthly_sales)


# 8. Monthly Sales Trend Graph
plt.figure(figsize=(12, 6))
plt.plot(monthly_sales.index, monthly_sales.values)
plt.title('Monthly Sales Trend')
plt.xlabel('Month')
plt.ylabel('Sales')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()


# 9. Category Sales Chart
plt.figure(figsize=(8, 5))
category_sales.plot(kind='bar')
plt.title('Sales by Category')
plt.xlabel('Category')
plt.ylabel('Sales')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()


# 10. Region Sales Chart
plt.figure(figsize=(8, 5))
region_sales.plot(kind='bar')
plt.title('Sales by Region')
plt.xlabel('Region')
plt.ylabel('Sales')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()


# 11. Top 10 Products Chart
plt.figure(figsize=(10, 6))
top_products.sort_values().plot(kind='barh')
plt.title('Top 10 Products by Sales')
plt.xlabel('Sales')
plt.ylabel('Product')
plt.tight_layout()
plt.show()


# 12. Sales vs Profit
plt.figure(figsize=(8, 6))
sns.scatterplot(data=df, x='Sales', y='Profit')
plt.title('Sales vs Profit')
plt.xlabel('Sales')
plt.ylabel('Profit')
plt.tight_layout()
plt.show()


# 13. Loss-Making Products
loss_products = (
    df.groupby('Product Name')['Profit']
    .sum()
    .sort_values()
)

print("\n===== 10 LOWEST-PROFIT PRODUCTS =====")
print(loss_products.head(10))


# 14. Discount Analysis
discount_profit = (
    df.groupby('Discount')['Profit']
    .mean()
    .sort_index()
)

print("\n===== AVERAGE PROFIT BY DISCOUNT =====")
print(discount_profit)

plt.figure(figsize=(8, 5))
discount_profit.plot(kind='bar')
plt.title('Average Profit by Discount')
plt.xlabel('Discount')
plt.ylabel('Average Profit')
plt.tight_layout()
plt.show()
```

```python
# Convert Month from Period to a normal string
df['Month'] = df['Month'].astype(str)

# Connect DuckDB
import duckdb

con = duckdb.connect()

# Register dataframe
con.register('sales_data', df)

print("DuckDB connected successfully!")
```

```python
# 1. Total Sales
print("===== TOTAL SALES =====")
print(con.sql("""
SELECT SUM(Sales) AS total_sales
FROM sales_data
"""))

# 2. Total Profit
print("\n===== TOTAL PROFIT =====")
print(con.sql("""
SELECT SUM(Profit) AS total_profit
FROM sales_data
"""))

# 3. Sales by Category
print("\n===== SALES BY CATEGORY =====")
print(con.sql("""
SELECT Category,
       SUM(Sales) AS total_sales
FROM sales_data
GROUP BY Category
ORDER BY total_sales DESC
"""))

# 4. Profit by Region
print("\n===== PROFIT BY REGION =====")
print(con.sql("""
SELECT Region,
       SUM(Profit) AS total_profit
FROM sales_data
GROUP BY Region
ORDER BY total_profit DESC
"""))

# 5. Top 10 Products
print("\n===== TOP 10 PRODUCTS =====")
print(con.sql("""
SELECT "Product Name",
       SUM(Sales) AS total_sales
FROM sales_data
GROUP BY "Product Name"
ORDER BY total_sales DESC
LIMIT 10
"""))

# 6. Sales by Segment
print("\n===== SALES BY SEGMENT =====")
print(con.sql("""
SELECT Segment,
       SUM(Sales) AS total_sales
FROM sales_data
GROUP BY Segment
ORDER BY total_sales DESC
"""))

# 7. Average Discount
print("\n===== AVERAGE DISCOUNT =====")
print(con.sql("""
SELECT AVG(Discount) AS average_discount
FROM sales_data
"""))

# 8. Loss-Making Products
print("\n===== LOSS-MAKING PRODUCTS =====")
print(con.sql("""
SELECT "Product Name",
       SUM(Profit) AS total_profit
FROM sales_data
GROUP BY "Product Name"
HAVING SUM(Profit) < 0
ORDER BY total_profit
LIMIT 10
"""))
```

```python
# Create a summary table
summary = pd.DataFrame({
    'Metric': [
        'Total Sales',
        'Total Profit',
        'Total Quantity',
        'Average Discount'
    ],
    'Value': [
        df['Sales'].sum(),
        df['Profit'].sum(),
        df['Quantity'].sum(),
        df['Discount'].mean()
    ]
})

print(summary)

# Save as Excel
summary.to_excel('sales_summary.xlsx', index=False)

print("\nExcel file created successfully!")
```
